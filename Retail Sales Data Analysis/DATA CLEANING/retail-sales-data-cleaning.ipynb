# 7. Data Cleaning & Preprocessing

## 7.1 Remove Duplicate Records

```python
df.drop_duplicates(inplace=True)
```

## 7.2 Handle Invalid Values

```python
df.loc[(df["age"]<0)|(df["age"]>100),"age"]=np.nan

non_negative_c = [
    "quantity",
    "unit_price",
    "discount_pct",
    "sales_amount",
    "shipping_cost",
    "days_to_ship",
    "customer_satisfaction"
]

for col in non_negative_c:
    df.loc[df[col] < 0, col] = np.nan
```

## 7.3 Handle Missing Values

```python

df["age"]=df["age"].fillna(df["age"].mean()).astype(int)
numeric_columns=[
"unit_price","sales_amount","profit","shipping_cost","quantity","discount_pct","customer_satisfaction","days_to_ship"
]
for col in numeric_columns:
    df[col]=df[col].fillna(df[col].median())
print(df.describe())
```

## 7.4 Clean Identifier Columns

```python
# Remove prefixes
df["order_id"] = (
    df["order_id"]
    .astype(str)
    .str.replace("ORD-", "", regex=False)
)

df["customer_id"] = (
    df["customer_id"]
    .astype(str)
    .str.replace("CUST", "", regex=False)
)

# Convert to numeric (invalid values become NaN)
df["order_id"] = pd.to_numeric(df["order_id"], errors="coerce")
df["customer_id"] = pd.to_numeric(df["customer_id"], errors="coerce")

# Drop rows where either ID is missing
df.dropna(subset=["order_id", "customer_id"], inplace=True)
```

## 7.5 Standardize Text Columns

```python

object_columns = ["customer_name","gender","region","city","product_category","product_name","payment_method","return_flag","order_status"]

for col in object_columns:
    df[col]=(df[col].astype(str).str.strip().str.title())
```

```python
df["gender"]=df["gender"].replace({
"male":"Male",
"MALE":"Male",
"m":"Male",
"M":"Male",
"female":"Female",
"FEMALE":"Female",
"f":"Female",
"F":"Female",
"other":"Other",
"OTHER":"Other"
})
print(df["gender"].value_counts())
```

## 7.6 Convert Data Types

```python
df["order_id"] = df["order_id"].astype(int)
df["customer_id"] = df["customer_id"].astype(int)
df["customer_satisfaction"] = pd.to_numeric(df["customer_satisfaction"], errors="coerce").astype(int)
df["days_to_ship"] = pd.to_numeric(df["days_to_ship"], errors="coerce").astype(int)
df["quantity"] = pd.to_numeric(df["quantity"], errors="coerce").astype(int)
df["order_date"] = pd.to_datetime(df["order_date"],format="mixed",dayfirst=True,errors="coerce")
```
