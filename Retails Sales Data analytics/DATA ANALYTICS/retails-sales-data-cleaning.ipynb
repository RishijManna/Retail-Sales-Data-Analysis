# 1. Customer Demographics Analysis

Understanding customer demographics helps businesses identify their target audience and develop effective marketing strategies. This section analyzes the distribution of customers based on their age and gender.

## 1.1 Customer Age Distribution

### Objective

To analyze the distribution of customer ages and identify the dominant age groups within the retail customer base.

```python
import plotly.express as px

fig = px.histogram(
    df,
    x="age",
    nbins=15,
    marginal="box",
    title="Customer Age Distribution",
    opacity=0.85,
    color_discrete_sequence=["#1f77b4"]
)

fig.update_layout(
    template="plotly_white",
    xaxis_title="Age",
    yaxis_title="Number of Customers",
    hovermode="x unified",
    title_x=0.5
)

fig.show()
```

### Observation

- The age distribution illustrates how customers are spread across different age groups.
- The histogram helps identify the most common customer age range.
- The KDE curve provides an overview of the overall distribution pattern.
- These insights help businesses understand the dominant customer demographic.

## 1.2 Customer Gender Distribution

### Objective

To analyze the gender composition of customers and determine the proportion of male, female, and other customers.

```python
gender_counts = (
    df["gender"]
    .value_counts()
    .reset_index()
)

gender_counts.columns = ["Gender", "Count"]

fig = px.pie(
    gender_counts,
    names="Gender",
    values="Count",
    hole=0.4,
    title="Customer Gender Distribution"
)

fig.update_traces(
    textposition="inside",
    textinfo="percent+label",
    hovertemplate="<b>%{label}</b><br>Customers: %{value}<br>Percentage: %{percent}"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()
```

### Observation

- The pie chart illustrates the proportion of customers across different gender categories.
- It helps identify the dominant customer segment.
- Understanding gender distribution assists in designing targeted marketing campaigns and customer engagement strategies.

## 4.2 Monthly Sales Trend

```python
print(df["order_date"].dtype)
```

```python
df["order_date"] = pd.to_datetime(
    df["order_date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)
```

```python
print(df["order_date"].dtype)
```

```python
print(df["order_date"].isna().sum())
```

```python
monthly_sales = (
    df.groupby(df["order_date"].dt.to_period("M"))["sales_amount"]
      .sum()
      .reset_index()
)

monthly_sales["order_date"] = monthly_sales["order_date"].astype(str)

fig = px.line(
    monthly_sales,
    x="order_date",
    y="sales_amount",
    markers=True,
    title="Monthly Sales Trend"
)

fig.update_traces(
    hovertemplate="<b>%{x}</b><br>Total Sales: ₹%{y:,.2f}<extra></extra>"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Month",
    yaxis_title="Total Sales (₹)",
    hovermode="x unified"
)

fig.show()
```

## 5.4 Product Category vs Sales Amount and Profit

### Objective

The objective of this analysis is to compare the total sales amount and total profit generated by each product category. This visualization helps identify product categories that generate high revenue as well as high profitability, enabling better inventory planning and business decision-making.

```python
import plotly.express as px

category_summary = (
    df.groupby("product_category", as_index=False)
      .agg({
          "sales_amount": "sum",
          "profit": "sum"
      })
)

# Convert to long format for grouped bar chart
category_summary = category_summary.melt(
    id_vars="product_category",
    value_vars=["sales_amount", "profit"],
    var_name="Metric",
    value_name="Amount"
)

# Rename legend labels
category_summary["Metric"] = category_summary["Metric"].replace({
    "sales_amount": "Sales Amount",
    "profit": "Profit"
})

fig = px.bar(
    category_summary,
    x="product_category",
    y="Amount",
    color="Metric",
    barmode="group",
    text_auto=".2s",
    title="Product Category vs Sales Amount and Profit"
)

fig.update_traces(
    hovertemplate="<b>%{x}</b><br>%{fullData.name}: ₹%{y:,.2f}<extra></extra>"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Product Category",
    yaxis_title="Amount (₹)",
    legend_title="Metric",
    hovermode="x unified"
)

fig.show()
```

### Observation

- The grouped bar chart compares the total sales amount and total profit generated by each product category.
- Hovering over the bars displays the exact sales and profit values for each category.
- Categories with both high sales and high profit represent the strongest performing product segments.
- Differences between sales and profit help identify categories with lower profit margins despite generating high revenue.
